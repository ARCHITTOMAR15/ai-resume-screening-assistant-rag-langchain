## Step 1 Reading PDF Resume

```python
import warnings 

from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader
```

```python
# create path for resume folder 
resume_folder=Path(r"C:\Users\Lenovo\Desktop\github projects\gen ai agentic ai assingment\data\resumes")

pdf_files=list(resume_folder.glob("*.pdf"))

print(f"Total Resume Found : {len(pdf_files)}")
```

```python
# print file names
for pdf in pdf_files:
    print(pdf.name)
```

```python
# Convert it to documnets format for Lnagchain 
```

```python
documents=[]
for pdf in pdf_files:
    loader=PyPDFLoader(str(pdf))
    documents.extend(loader.load())
```

```python
print(f"Total Documents Loaded: {len(documents)}")
```

```python
documents[0]
```

```python
documents[0].page_content[:1000]
```

```python
documents[0].metadata
```

```python
# LETS ENRICH OUR METS DATA
for doc in documents:
    resume_file=Path(doc.metadata["source"]).name
    doc.metadata["resume_file"]=resume_file
    doc.metadata["candidate_name"]=Path(resume_file).stem
```

```python
for i,doc in enumerate(documents[0:4]):
    print(f"Documents{i+1}")
    print(doc.metadata)
    print("+"*50)
```

#### TEXT CHUNKING

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter= RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=100,separators=["\n\n","\n",". "," ",""])
```

```python
chunked_documents=text_splitter.split_documents(documents)
```

```python
print(f"Total Chunk Created {len(chunked_documents)}")
```

```python
for i ,chunk in enumerate(chunked_documents[0:3]):
    print(f"Documents {i+1}")
    print("+"*50)
    print (chunk.metadata)
    print("+"*50)
    print(chunk.page_content)
    print("="*50)
```

#### Chunk Quality Analysis

```python
# Total Chunks 
print(f"Total Chunks {len(chunked_documents)}")
```

```python
# Total Chunk per resume

from collections import Counter

resume_chunk_count = Counter(
    chunk.metadata["resume_file"]
    for chunk in chunked_documents
)

print("Chunks per Resume:\n")

for resume, count in resume_chunk_count.items():
    print(f"{resume}: {count} chunks")
```

```python
# Chunk Lenghth Stasticss 
chunk_lengths=[]

for chunk in chunked_documents:
    chunk_length=len(chunk.page_content)
    chunk_lengths.append(chunk_length)

print(f"Maximum Length of Chunk: {max(chunk_lengths)}")
print(f"Minimum Length of Chunk:{min(chunk_lengths)}")
print(f"Average Length of Chunk Length:{sum(chunk_lengths)/len(chunk_lengths)}")
```

```python
# Inspect Random Chunk 
import random 
sample_chunk=random.sample(chunked_documents,min(3,len(chunked_documents)))

for i,chunk in enumerate(sample_chunk,start=1):
    print(f" sample_chunk {i}")
    print(chunk.metadata)
    print("_"*50)
    print(chunk.page_content)
    print("+"*50)
```
