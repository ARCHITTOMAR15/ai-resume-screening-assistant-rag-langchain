## 9. RAG CHAIN

### 9.1 JD + Resume Evaluation Prompt

```python
evaluation_prompt=ChatPromptTemplate.from_template("""
You are an AI Resume Screening Assistant.

Evaluate the candidate against the provided Job Description.

Use ONLY the information contained in the retrieved resume context.

IMPORTANT RULES:
- Do not invent candidate skills or experience.
- Do not assume a skill is present unless supported by the resume.
- If a required skill is not found in the resume context,
  consider it a missing skill.
- Base the evaluation only on the provided resume context.
- Be objective and factual.

JOB DESCRIPTION:
{job_description}

RESUME CONTEXT:
{context}

Evaluate the candidate against the Job Description.
""")
```

```python
# lets test 
job_description = """
Data Scientist required with:
Python
Machine Learning
SQL
Pandas
Scikit-learn
TensorFlow
Docker
"""

```

### 9.2 OUTPUT SCHEMA 

```python
from pydantic import BaseModel, Field
```

```python
from typing import Literal
class ResumeEvaluation(BaseModel):
    match_score: int = Field(description="Score from 0 to 100")
    matching_skills: list[str]
    missing_skills: list[str]
    recommendation: Literal["Hire", "Shortlist", "Reject"]
```

```python
# output Parser
from langchain_core.output_parsers import PydanticOutputParser

parser = PydanticOutputParser(
    pydantic_object=ResumeEvaluation
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

structured_prompt = ChatPromptTemplate.from_template("""
You are a resume screening assistant.

Compare the Job Description with the Resume Context.

Use ONLY the resume context.

IMPORTANT:

1. matching_skills:
   Include ONLY skills that are explicitly present
   in BOTH the Job Description and Resume Context.

2. missing_skills:
   Include ONLY skills that are present in the Job Description
   but are NOT present in the Resume Context.

3. A skill MUST NOT appear in both lists.

4. Do not invent information.

5. match_score must be between 0 and 100.

6. recommendation must be one of:
   "Hire"
   "Shortlist"
   "Reject"

Return only the requested structured output.

Job Description:
{job_description}

Resume Context:
{context}

{format_instructions}
""")
```

## 10. Resume Screening and Candidate Evaluation

```python
# specific chain with littile change 

candidate_chain = (
    {
        "context": lambda x: x["context"],
        "job_description": lambda x: x["job_description"],
        "format_instructions": lambda _: parser.get_format_instructions()
    }
    | structured_prompt
    | llm
    | parser
)
```

```python
## function for specific candidate
```

```python
# Creatinglist of resumes 
resume_names = [
    doc.metadata["resume_file"]
    for doc in chunked_documents
]

resume_names = list(dict.fromkeys(resume_names))

print(resume_names)
```

#### 2. Evaluate multiple resumes

```python
#Evaluatimg every resume 

candidate_results = {}

for resume_name in resume_names:

    resume_chunks = [
        doc for doc in chunked_documents
        if doc.metadata["resume_file"] == resume_name
    ]

    resume_context = format_doc(resume_chunks)

    result = candidate_chain.invoke({
        "job_description": job_description,
        "context": resume_context
    })

    candidate_results[resume_name] = result
```

```python
for resume_name, result in candidate_results.items():

    print("=" * 60)
    print("CANDIDATE:", resume_name)
    print("=" * 60)

    print("Match Score:", result.match_score)
    print("Matching Skills:", result.matching_skills)
    print("Missing Skills:", result.missing_skills)
    print("Recommendation:", result.recommendation)
    print()
```

#### 3. Rank candidates

```python
## create ranking 
ranked_candidates = sorted(
    candidate_results.items(),
    key=lambda x: x[1].match_score,
    reverse=True
)
```

```python
# display ranking 

for rank, (resume_name, result) in enumerate(ranked_candidates, start=1):

    print(
        f"{rank}. {resume_name} "
        f"| Score: {result.match_score} "
        f"| Recommendation: {result.recommendation}"
    )
```

#### 4. Show best candidate

```python
# get the best candidate


best_candidate = ranked_candidates[0]

print("Best Candidate:", best_candidate[0])
print("Match Score:", best_candidate[1].match_score)
print("Recommendation:", best_candidate[1].recommendation)
```
